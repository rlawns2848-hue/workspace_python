```python
import pandas as pd

df = pd.read_excel('data.xlsx')
df
```

```python
# 결측치가 아닌 데이터 변환 replace()

# replace 사용 후 결과를 저장하려면 inplace=True 매개변수를 추가하거나 다른 변수에 저장해야 한다.
# 모든 데이터(컬럼 구분 안 함)에서 '응시'문자열을 'Y'로 변환 잘 안 씀
df.replace('응시','Y') 
#df.replace('응시','Y', inplace=True) 

# 모든 데이터에서 '남','여' 문자열을 변경한다.
df.replace({'남':'male', '여':'female'})

# 특정 컬럼의 값을 변경
df['확인여부'].replace('확인', '1')
df
```

```python
# 컬럼 추가
# 컬럼 추가, 제거는 inplace=True 대상이 아니다.
# 테이블 구조 자체가 변하는 것은 무조건 저장이 된다.
# 총점 컬럼 추가 및 국어와 수학의 합을 데이터로 삽입
# 없는 컬럼명을 넣어주면 추가된다.
df['총점'] = df['국어'] + df['수학']
#함수가 아니라서 바로 저장
df
```

```python
# 컬럼 추가, assign() 함수 사용
# assign() 함수는 컬럼 추가한 결과를 반환해준다.
# 원본은 유지하면서, 컬럼이 추가된 데이터셋을 사용할 수 있다.
# 변수를 원본데이터로 주면 원본데이터를 변결할 수 있다.

# 평균 컬럼 추가
new_df = df.assign(평균=df['총점']/2)
new_df # 평균 컬럼이 있다.
df # 평균 컬럼이 없다.
```

```python
# 데이터 삭제 drop(), 저장하려면 inplace=True 주거나 다른 변수에 저장하거나 해야한다.
# inplace=True는 함수로 만들어진 것은 거의 다 쓸 수 있다고 생각하자. 
# axis='index', axis=0 -> 행 삭제
# axis='columns', axis=1 -> 열 삭제

# 총점 컬럼 삭제
df.drop(['총점'], axis='columns')

# 국어, 수학 컬럼 삭제
df.drop(['국어','수학'], axis='columns')

# 0, 1번 째 행을 삭제
df.drop([0, 1], axis='index')

# 2번 째 행 삭제
df.drop([2], axis=0)
```
