```python
#리턴이 있는 함수 
#한 번 실행해야 만들어짐
def getSum():
  print('두 수의 합을 리턴합니다.')
  return 10 + 20
```

```python
result1 = getSum()
print(result1)
```

```python
#파이썬은 여러 데이터를 리턴할 수 있다.
#return은 한 번만 쓸 수 있지만 ,로 여러 데이터를 리턴할 수 있다
def multiReturn():
  return 10, 20
```

```python
#리턴이 여러개인 함수를 받으려면 변수도 여러 개 만들어서 받는다.
#순서대로 들어온다
result2, result3 = multiReturn()
print(result2)
print(result3)
```

```python
#매개변수가 있는 함수
#자료형이 없으니 적지 않는다
def getSub(a,b):
  return a - b
```

```python
result4 = getSub(10,1)
print(result4)
```

```python
#파이썬은 매개변수에 이름을 주고, 이름을 사용하여 데이터를 전달할 수 있다.
#보통 매개변수에 넣은 순서대로 들어가지만, 이름으로 직접 지정할 수 있다. 
result5 = getSub(b = 10, a = 20)
print(result5)
```

```python
#파이썬은 함수 정의 시 매개변수에 default 값을 줄 수 있다.
#함수 선언문에서 변수에 값을 넣고 호출할 때 변수에 값을 넣지 않으면 20이 디폴트값으로 들어온다.
def tellMyIntro(name, age = 20):
  print(f'내이름은 {name}이고, 나이는 {age}살입니다.')
```

```python
tellMyIntro('kim', 30)
#매개변수에 값을 안 넣어도 default값이 있으니 그 값으로 나온다
tellMyIntro('lee')
tellMyIntro(age = 40, name = 'hong')
```

함수 실습

```python
#1.아래 주어진 딕셔너리 데이터를 매개변수로 받으면, 모든 점수의 합을 리턴하는 힘수 선언 및 호출
score_dic = {
  'kor' : 70
  , 'eng' : 75
  , 'math' : 90
}
```

```python
def totalScore(score_dic):
  return score_dic['kor'] + score_dic['eng'] + score_dic['math']

print(totalScore(score_dic))

#풀이
#함수 정의
def test1(dic_data):
  return dic_data['kor'] + dic_data['eng'] + dic_data['math']

#함수 호출
test1(score_dic)
```

```python
#2. 첫 번째 매개변수로 전달되는 딕셔너리 데이터에 두 번째 매개변수로 전달된 key와
# 세번째 매개변수로 전달된 value값을 추가한 딕셔너리 데이터를 리턴하는 함수를 선언 및 호출  
```

```python
def addTotal(score_dic, b1, c1):
  score_dic[b1] = c1

addTotal(score_dic,'total', totalScore(score_dic))
print(score_dic)  

#풀이
def test2(dic_data, key, value):
  dic_data[key] = value
  return dic_data

person = { 
  'name' : 'kim'
  , 'age' : 20
}

test2_result = test2(person, 'addr', '울산시')
print(test2_result)
```

```python
#3. 매개변수로 문자열을 갖는 리스트가 전달되면 리스트의 문자열과 문자열의 길이를 
# 각각 key와 value로 받는 딕셔너리를 리턴하는 함수 선언 및 호출
# ex
# 매개변수 : ['apple','banana','orange']
# 리턴데이터
# result = {'apple' : 5,'banana' : 6,'orange' : 6}
```

```python
fruit =['apple','banana','orange', 'mango','strawberry']
basket = {}
def makeDic( dic_data):
  for e in dic_data:
    basket[e] = len(e)

makeDic( fruit)

print(basket)

#풀이
#len() 리스트의 길이를 알려줌
#파이썬은 문자열을 리스트로 취급
def test3(str_list):
  #결과 데이터
  result = {}
  # 매개변수로 전달된 리스트 데이터에 하나하나에 접근
  for e in str_list:
    result[e] = len(e)
  return result # for문 밖이면서 함수 안에 있어야하니 위치 조심

test3_result = test3(['apple','banana','orange'])  
print(test3_result)
```
