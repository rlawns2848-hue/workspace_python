파이썬의 출력문

```python
# 파이썬은 명령어 다음에 세미콜론(;) 작성하지 않는다.
print(1)
print("python")
print('python')
print(10 + 3)
print('a'+'b')
#print('a'+7)파이썬은 문자열과 숫자 더하기 연산이 불가하다!
print(10/4) #소수점 결과까지 표현한다.
print(10%4)

#파이썬에만 있는 문법
print(10//4) #몫 계산

#또 다른 출력 문법
print(10,5,'a')
```

파이썬의 변수


```python
num = 5 # 최초 사용하는 변수는 자동으로 변수를 생성한다.
num = 10 # num변수의 값을 10으로 변경한다.
num = 'aaaa' #자료형 상관없이 데이터 저장이 가능하다.

myName = 'Kim'

#변수 출력
print(myName)
print(num)
```

```python
#셀이 달라도 실행한 내용은 전부 연결되어 있다.
print(num)
```
