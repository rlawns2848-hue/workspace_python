## 딕셔너리(dictionary)
자바스크립트의 객체 자료형과 동일하다.  
key와 value를 한 쌍으로 다수의 데이터를 가질 수 있는 자료형

```python
#마크다운 셀에서 한줄 개행하고 싶으면 문장 끝에 띄어쓰기 2번한다
#빈 딕셔너리 데이터를 생성
dic1 = {}

#데이터가 있는 딕셔너리 생성
dic2 = {
  'name' : 'kim'
  ,'age' : 20
}

#전체 데이터 출력
print(dic2)
#일부 데이터 출력 배열처럼 [key]로추출
print(dic2['name'])
```

```python
#딕셔너리 데이터 추가 및 수정
phone = {
  'modelName' : 's20'
  , 'price' : 10000
}

#데이터 추가 
#딕셔너리가 가지지 않은 key를 사용하면 데이터가 추가
phone['color'] = 'red'
print(phone)

#존재하지 않는 key값의 데이터를 읽고 출력할 수는 없다.
#print(phone[brand])

#데이터 수정 
#기존 key에 데이터를 저장하면 수정된다.
phone['price'] = 20000
print(phone)

#데이터 삭제
#하나의 key값을 삭제
del phone['modelName'] 
print(phone)
#딕셔너리 데이터 변수 자체를 삭제
#del phone
```
