## 반복문

```python
#일반 반복문

#for-each문과 비슷
#1~10까지 출력
#range(시작숫자, 마지막에 포함되지 않는 수)
for e in range(1,11) : 
  print(e)
```

```python
#10~30까지의 정수 중에서 짝수만 출력
for e in range(10,31) : 
  if(e % 2 == 0) :
    print(e)
```

```python
#파이썬은 변수명 사용 시 주의가 필요하다.
#변수는 약속된 자료형은 변수로 쓸 수 없다 ex) String int; X
#변수명을 기존에 존재하는 내장 함수명과 동일한 이름으로 변수를 사용하게되면 내장 함수가 동작하지 않게 된다.
#파이썬에 함수sum이 있기 때문에 변수로 쓰면 작동은 하지만 본래 함수로는 쓸 수 없게 된다.
#대표적 함수 : sum(), max(), min(), len(), list(), dic()
#실수로 내장된 함수와 동일한 이름으로 변수를 만들었다면 해당 변수는 삭제해야 한다.
#1~10까지의 합 출력
#이 문제를 풀기 위해 sum 변수를 생성 -> 문제가 발생할 수 있는 코드
sum_result = 0
for e in range(1,11) : 
  sum_result += e

print(sum_result)
```

```python
#sum변수 사용에는 문제 없지만 sum() 함수가 호출되지 않는 문제가 발생한다.
print(sum_result)

list1 = [1,2,3]
print(sum(list1))
```

```python
#위 셀에서 발생한 문제 해결을 위해 sum 변수를 삭제
del sum
```

```python
#for-each
list1 = [1,2,3,4,5]
for e in list1 :
  print(e)

for e in [1,2,3] :
  print(e)
```

```python
#반복 index 데이터
#반복 index를 사용할 수 있는 for문
#for index변수, 데이터 하나하나를 의미하는 변수 in enumerate(리스트 변수) :
list1 = [2,4,6,8,10]
#a = index, b = 리스트의 데이터
for a, b in enumerate(list1) :
  print(f'a = {a}, b = {b}')
```
