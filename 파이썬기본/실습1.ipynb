```python
#1. 1부터 키보드로 입력받은 정수까지의 합을 출력
#input로 입력받은 숫자는 문자로 친다.
#int('3') = 3
input_num = int(input('정수 입력 : ')) + 1
sum_result = 0
for e in range(1, input_num) :
  sum_result += e

print(sum_result)  
```

```python
#2. 키보드로 입력받은 단수의 구구단을 출력
#ex) 키보드로 3을 입력받으면 
# 3 * 1 = 3
# 3 * 2 = 6
# ...
# 3 * 9 = 27
num1 = int(input('구구단 단수 입력 : '))
for e in range(1,10) :
  print(f'{num1} * {e} = {num1 * e}')
```

```python
#3. 아래 제시된 리스트에서 가장 큰 요소와 가장 작은 요소의 차를 출력
sample_list = [3, 1, 7, 10, 5]

print(max(sample_list) - min(sample_list))
```

```python
#4. 아래 제시된 두 리스트의 모든 요소의 합을 출력
list1 = [1, 3, 5]
list2 = [10, 15, 20]
list3 = list1 + list2 # 리스트를 더하면 나열된다

print(sum(list1) + sum(list2))
print(sum(list3))
```

```python
#5. 아래 데이터에서 't'의 갯수를 출력
#hint : 파이썬은 문자열도 리스트로 취급
#파이썬은 증감연산자 없음
#name = 'hong' name[1] = 'o'
data = ['mother', 'father', 'brother']

tCnt = 0
for word in data : 
  #print(word)
  for c in word :
    #print(c)
    if c == 't':
      tCnt += 1

print(tCnt)
```

```python
#아래 제시된 딕셔너리 데이터를 참고하여 6~7번 문제 푸세요
#실행을 해야 변수가 만들어진다. 한번 실행할 것
student = {
  'name' : 'hong'
  , 'korScore' : 0
  , 'mathScore' : 0
}
```

```python
#6. student 딕셔너리에 키보드로 입력받은 데이터로 국어, 수학 점수를 변경하시오. 
korScore = int(input('국어 점수 :'))
mathScore = int(input('수학 점수 :'))

student['korScore'] = korScore
student['mathScore'] = mathScore

print(student)
```

```python
#7. student 딕셔너리에 key값으로 총점(totalScore)을 추가하고, value로는 국어, 수학 점수의 합을 저장하시오.
student['totalScore'] = student['korScore'] + student['mathScore']

print(student)
```
