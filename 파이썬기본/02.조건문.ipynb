## 조건문
파이썬은 세미콜론 없다. 동시에 중괄호도 없다.

```python
# if 조건 :
# 파이썬은 중괄호 대신 들여쓰기로 영역을 구분한다.
# 들여쓰기 잘 맞춰야한다. 
if 10 < 2 :
  print(1)
print(10)
```

```python
#키보드로 입력받은 정수가 짝수인지 홀수인지 구분하기
#키보드로 입력받은 데이터는 무조건 문자열로 인식하다.
#문자열을 숫자로 변경 : int('10')
#숫자 -> 문자열 : str(10)
num = int(input('정수 입력 : '))

if num % 2 == 0 :
  print('짝수')
else :
  print('홀수')  
```

```python
#키보드로 입력받은 정수가 양수인지 홀수인지 판단
num_1 = int(input('정수 입력 : '))

if num_1 > 0 :
  print('양수')
elif num_1 < 0 :
  print('음수')  
else : 
  print(0)
```

```python
#여러 조건 만족하기
if 10 > 3 and 5 > 4 :
  print('동시만족')

if 10 > 3 or 5 < 4 :
  print('하나만 만족')
```
