## 리스트
파이썬의 리스트는 자바스크립트의 배열과 기본적인 동작이 같다.  
다만, 파이썬에서는 리스트의 내요을 컨트롤하는 다양한 기능들이 있다

```python
#빈 리스트 생성
list1 = []

#데이터를 저장하고 있는 리스트 생성
list2 = [1,2,3.5,'pyhton']
print(list1)
print(list2)
print(list2[1])
```

```python
list3 = [1,2,3]

#리스트에 데이터 추가
#1. append() : 리스트의 마지막 요소에 데이터 추가
list3.append(4)
print(list3)
#2. insert(원하는 자릿수, 추가할 데이터) : 리스트의 원하는 요소에 데이터 추가
list3.insert(2,5) #리스트의 2번째 요소에 5 추가
print(list3)
```

```python
#리스트 데이터 삭제
list5 = [1,2,3,4,5]

#1번 요소 삭제
del list5[1]
print(list5)

del list5[2]
print(list5)

del list5 #list5 변수 자체를 삭제
#print(list5) 윗 코드에서 list5 변수를 삭제했기 때문에 오류가 발생한다.
```

```python
#파이썬에만 있는 문법
list1 = [1,2,3]
list2 = [4,5,6]
#위 두 리스트를 나열(합치고 싶다.) -> [1,2,3,4,5,6]
list3 = list1 + list2
print(list3)

list4 = [1,2,3]
#[1,2,3,1,2,3,1,2,3]
list5 = list4 * 3
print(list5)
```

```python
#슬라이싱 연산(리스트에서 일부 데이터만 추출)
#list[시작index : 마지막 index(포함안됨)] 마지막 index 전까지 추출된다.
#list의 자릿수는 왼쪽부터 시작 : 0,1,2,3,4, ~
#               오른쪽부터 시작 : ~ ,-3,-2,-1 
list1 = [1,2,3,4,5]
print(list1[2:4]) #2번index ~ 3번 index [3,4]
print(list1[1:]) # 1번index ~ 끝까지 [2,3,4,5]
print(list1[:3]) # 첫index ~ 2번index까지 [1,2,3]
print(list1[:]) # 처음부터 끝까지
print(list1[1:-2]) #[2,3]

#파이썬은 문자열도 리스트로 인식한다.

address = 'python'
print(address[0])
print(address[0:2]) #py
```
